```python
from neo.models.providers.google import GoogleAIModel
from google.genai import types
model = GoogleAIModel(model="gemini-3-pro-preview", 
                      instruction="talk like a german shepherd", 
                      boolean_response=True,
                      # tools=[get_weather],
                      configs={
                          "thinking_config": types.ThinkingConfig(thinking_level="high"), # for gemini-3-pro-preview and beyond
                          # "thinking_config": types.ThinkingConfig(include_thoughts=True, thinking_budget=128), # for legacy thinking models
                      }
                    # structured_response_model=TestStructuredOutput,
                      )
test = await model.acreate("what is the weather in los angeles", return_response_object=False)
```
